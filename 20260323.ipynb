```python
eng = 90
math = 100

if eng >= 90 and math >= 90 :
    print ('스티커')
else :
    print ('좀 더 노력하세요')
```

```python
둘다 90점 이상이면 둘다성공
영어는 90점 이상인데 수학은 그렇지 않으면 좀 더 노력
둘다 90점 미만이면 많이 노력
```

```python
eng = 90
math = 100

if eng >= 90 and math >= 90 :
    print ('둘 다 성공')
elif eng >= 90 and math < 90 :
    print ('좀 더 노력하세요')
elif eng < 90 and math < 90 :
    print ('많이 노력하세요')
```

```python
# 인선미 강사 풀이

eng = 90
math = 90

if eng >= 90 :
    if math >= 90 :
        print ('둘 다 성공')
    else :
        print ('좀 더 노력')
else :
    print ('많이 노력')
```

```python
# 미션 :  성적 처리 프로그램
'''
90점 이상 100점 미만 A학점
80점 이상 90점 미만 B학점
70점 이상 80점 미만 C학점
60점 이상 70점 미만 D학점
60점 미만 F학점
'''

a = int(input('점수 입력 : '))

if a >= 90 :
    print ('A학점')
elif a >= 80 :
    print ('B 학점')
elif a >= 70 :
    print ('C 학점')
elif a >= 60 :
    print ('D 학점')
elif a < 60 :
    print ('F 학점')
else :
    print ('종료')

## 순서를 지키지 않을시 꼬일 수 있다 
# 그럼으로 조건에 and를 사용하는것이 좋다
```
