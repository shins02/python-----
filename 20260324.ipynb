```python
# 도전 미션

'''
리스트 numbers 각 원소는 숫자 0부터 9까지로 이루어져 있습니다. 이때, numbers에서 연속적으로 나타나는 숫자는 하나만 남기고 전부 제거하려고 합니다. 단, 제거된 후 남은 수들은 numbers의 원소의 순서를 유지해야 합니다.
 
예를 들면,
numbers = [1, 1, 3, 3, 0, 1, 1] ->  [1, 3, 0, 1] 
numbers = [4, 4, 4, 3, 3] ->  [4, 3]
 
코드를 작성해 주세요
'''

numbers1 = [1, 1, 3, 3, 0, 1, 1] 


answer = []
comp = numbers1 [0]
answer.append (comp)

for i in numbers1:
    if comp != i:
        comp = i
        answer.append(comp)
answer 
```

```python
def num_max(a,b,c):
    max_val = a
    if b > max_val:
        max_val = b
    if c > max_val:
        max_val = c
        
    return max_val
```

```python
def num_min():
    num_1 = list(num_min(0))
    if list(num_min(1)) < num_1:
        num_1 = list(num_min(1))
    if list(num_min(2)) < num_1 :
        num_1 = list(num_min(2))

    return num_1
```

```python
#인선미 강사 예시

def num_min(arr):
    # print(arr)
    min_val = arr[0]
    for i in arr:
        if min_val > i :
            min_val = i

    return min_val
```

```python
num_min([4, 8,2 , 1])
```

```python
# 인선미 강사 예시

def add_mul(selection, * args):
    print(selection, args)
    if selection == 'add':
        resurt = 0
        for i in args :
            resurt += i
    elif selection =='mul':
        resurt = 1
        for i in args :
            resurt *= i
    else:
        print('연산자 종류 확인 할것!!')
        resurt = None
    return resurt
```
